```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python
ticker = "GOOGL"
stock = yf.Ticker(ticker)

price = stock.history(period="5y")["Close"]
price.name = "price"
```

```python
# Annual income statement
income = stock.income_stmt.T  # transpose: rows = years
eps = income["Diluted EPS"]

eps.index = pd.to_datetime(eps.index)
eps.name = "eps"
```

```python
df = pd.concat([price, eps], axis=1)
df["eps"] = df["eps"].ffill()
df.dropna(inplace=True)
```

```python
df["pe"] = df["price"] / df["eps"]
```
